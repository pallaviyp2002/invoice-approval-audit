```python
np.random.seed(42)
```

```python
df = pd.DataFrame({
    'invoice_id': [f'INV-{i}' for i in range(1, 201)],
    'vendor':     np.random.choice(['Vendor A','Vendor B','Vendor C','Vendor D'], 200),
    'amount':     np.round(np.random.uniform(10000, 300000, 200), 2),
    'department': np.random.choice(['HR','IT','Finance','Operations'], 200),
    'approved':   np.random.choice(['Yes','No'], 200, p=[0.85, 0.15])
})
```

```python
df.to_csv('invoices.csv', index=False)
print(df.head())
```

```python
import pandas as pd
df = pd.read_csv('invoices.csv')

high_value = df[df['amount'] > 100000]
failure_rate = (high_value['approved'] == 'No').mean()

print(f"Total high-value invoices : {len(high_value)}")
print(f"Unapproved                : {(high_value['approved']=='No').sum()}")
print(f"Failure Rate              : {failure_rate:.1%}")

# Is the failure rate statistically above 5% tolerance?
from scipy.stats import binomtest
unapproved_count = (high_value['approved']=='No').sum()
total_count = len(high_value)

result = binomtest(unapproved_count,n=total_count,p=0.05, alternative='greater')

p_value = result.pvalue


print(f"P-value                   : {p_value:.4f}")
print("CONTROL FAILED" if p_value < 0.05 else "Control OK")


```
